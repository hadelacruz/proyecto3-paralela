### 1.1 Operaciones del programa, de la imagen al resultado

| No. | Operación identificada | Descripción de su funcionamiento |

| 15 | Actualización Adam | Actualiza cada parámetro con sus momentos (1.1 millones de parámetros en la CNN Base). |
| 16 | Validación por época | Propagación hacia adelante sobre 3,480 imágenes al final de cada época. |
| 17 | Inferencia y métricas | `predict` sobre prueba, `argmax`, exactitud, matriz de confusión. |

````python
# 1.2 Cuántas veces se repite cada operación (conteos exactos para esta configuración)
def macs_conv(lado, c_in, c_out):
    return lado * lado * c_out * 9 * c_in

capas = {
    'CNN Base': [('Conv 3→32 @64×64', macs_conv(64, 3, 32)), ('Conv 32→64 @32×32', macs_conv(32, 32, 64)),
                 ('Conv 64→128 @16×16', macs_conv(16, 64, 128)), ('Densa 8192→128', 8192 * 128),
                 ('Densa 128→29', 128 * 29)],
    'CNN Profunda': [('Conv 3→32', macs_conv(64, 3, 32)), ('Conv 32→32', macs_conv(64, 32, 32)),
                     ('Conv 32→64', macs_conv(32, 32, 64)), ('Conv 64→64', macs_conv(32, 64, 64)),
                     ('Conv 64→128', macs_conv(16, 64, 128)), ('Conv 128→128', macs_conv(16, 128, 128)),
                     ('Densa 128→256', 128 * 256), ('Densa 256→29', 256 * 29)],
}
N_TRAIN, LOTE = 16_240, 64
for modelo, lista in capas.items():
    total = sum(m for _, m in lista)
    tabla_capas = pd.DataFrame([(n, m / 1e6, m / total) for n, m in lista],
                               columns=['Capa', 'MACs por imagen (millones)', '% del total'])
    print(f"{modelo}: {total / 1e6:.1f} M multiplicaciones-suma por imagen hacia adelante; "
          f"~{3 * 2 * total * N_TRAIN / 1e12:.1f} TFLOP por época (adelante + atrás ≈ 3×)")
    display(tabla_capas.style.format({'MACs por imagen (millones)': '{:.2f}', '% del total': '{:.1%}'}))
````

## 3. Identificación de oportunidades de paralelización

| Operación | ¿Paralelizable? | Dónde | Justificación |
|---|---|---|---|
| Lectura, decodificación y redimensión de JPEG | Sí, por imagen (23,200 tareas independientes) | **CPU, varios procesos** | Decodificar JPEG es lógica con muchas ramas (Huffman) y acceso a disco, para lo que un núcleo de CPU es mejor. En Python los hilos no escalan por el GIL, así que se usan procesos. |
| Normalización ÷255 | Sí, por elemento | CPU (dentro del mismo proceso) | Es trivial y va pegada a la decodificación: mandarla a la GPU costaría más en transferencia que lo que ahorra. |
| Partición y barajado | Poco trabajo | CPU | Opera sobre índices; es microsegundos. |
| Convoluciones (adelante y atrás) | Sí, por cada valor de salida (millones por lote) | **GPU** | Son el 89–98% del cómputo; operaciones aritméticas idénticas e independientes, ideales para miles de hilos. |
| Capas densas | Sí, producto de matrices | **GPU** | Productos de matrices en baldosas, con reutilización en memoria compartida. |
| ReLU, pooling, dropout, BatchNorm | Sí, por elemento o por canal | **GPU** | Elemento a elemento; se ejecutan donde ya están los datos para no transferirlos. |
| Softmax + pérdida | Sí, por imagen | **GPU** | 64 hilos por lote; los datos ya están en la GPU. |
| Actualización de Adam | Sí, por parámetro | **GPU** | 1.1 M de actualizaciones independientes por paso. |
| Aumento de datos | Sí, por imagen | **GPU** | En Keras las capas de aumento son parte del modelo; se conserva ese diseño y corren en la GPU como transformaciones afines por lote. |
| Validación e inferencia | Sí | **GPU** | Solo propagación hacia adelante; misma lógica que el entrenamiento. |
| Algoritmos clásicos (Random Forest, SVM, KNN) | — | CPU (sin cambios) | scikit-learn no usa GPU y cuML (RAPIDS) no está disponible para Windows; quedan fuera del alcance. |

**La idea central:** cada tipo de trabajo va al procesador que lo hace mejor. La CPU tiene pocos núcleos rápidos, buenos para lógica con ramas y E/S, y la GPU tiene cientos de núcleos simples, buenos para la misma operación aritmética sobre millones de datos. Las imágenes solo cruzan a la GPU ya convertidas en tensores, en lotes de 64 (3 MB por lote).

## 4. Diseño del algoritmo paralelo

### 4.1 Fase 1 en CPU: procesos con memoria compartida

```
proceso principal                                procesos trabajadores (8)
──────────────────                               ─────────────────────────
selecciona 23,200 rutas (semilla fija)
reserva SharedMemory de 1.14 GB  ───────────────►  cada uno se conecta al bloque por su nombre
divide en 32 tramos contiguos ──── tramo k ─────►  decodifica + redimensiona + ÷255
                                                   escribe X[inicio:fin] directo en el bloque
espera los 32 resultados (pid, inicio, n) ◄──────  devuelve solo 3 enteros
copia el bloque a un arreglo normal
```

* **Procesos, no hilos:** cada proceso tiene su propio intérprete y su propio GIL.
* **4 tramos por proceso (32 en total)** para balancear la carga: el que termina antes toma otro tramo.
* **Memoria compartida:** sin ella, cada proceso devolvería su parte serializada (pickle) y se copiarían 1.14 GB entre procesos. Así, cada proceso devuelve solo 3 números.

### 4.2 Fase 2 en GPU con PyTorch + cuDNN

* Los tensores de entrenamiento quedan en la RAM del CPU en memoria *pinned*. Cada lote de 64 imágenes (3 MB) se copia a la GPU por DMA; el conjunto completo (~800 MB) cabría en los 2 GB de la MX450, pero dejaría poco margen para activaciones y espacio de trabajo de cuDNN.
* cuDNN elige el algoritmo de convolución más rápido para cada forma de capa (`cudnn.benchmark = True`).
* La pérdida y los aciertos se acumulan en la GPU y se leen una vez por época: leerlos en cada lote obligaría a sincronizar CPU y GPU 254 veces por época.

### 4.3 Kernels propios en CUDA Python (Numba): hilos, bloques y grid

Para diseñar explícitamente el reparto del trabajo, la CNN Base completa (adelante, atrás y Adam) se implementó con kernels propios en `src/kernels_cuda.py`. La celda siguiente calcula la configuración real de lanzamiento de cada kernel para un lote de 64 imágenes.

````python
from src import kernels_cuda as kc
N = 64
filas = []
for i, ((ci, co), lado) in enumerate(zip(kc.CAPAS_CONV, (64, 32, 16)), 1):
    P = N * lado * lado
    filas += [
        (f'conv{i} adelante', 'k_conv_gemm_rt', f'({-(-P // kc.BM)}, {-(-co // kc.BN)})', '8×16 = 128',
         '4×4 salidas por hilo', -(-P // kc.BM) * -(-co // kc.BN) * 128),
        (f'conv{i} grad. pesos', 'k_conv_gemm_pesos',
         f'({-(-co // 16)}, {-(-9 * ci // 16)}, {-(-P // 4096)})', '16×16 = 256',
         'split-K: 4,096 píxeles por bloque', -(-co // 16) * -(-9 * ci // 16) * -(-P // 4096) * 256),
    ]
    if i > 1:
        filas.append((f'conv{i} grad. entrada', 'k_conv_gemm_rt (pesos volteados)',
                      f'({-(-P // kc.BM)}, {-(-ci // kc.BN)})', '8×16 = 128', '4×4 salidas por hilo',
                      -(-P // kc.BM) * -(-ci // kc.BN) * 128))
    Pp = P // 4
    filas.append((f'pool{i} adelante', 'k_maxpool_forward', f'({-(-co // 32)}, {-(-Pp // 8)})', '32×8 = 256',
                  '1 salida por hilo', -(-co // 32) * -(-Pp // 8) * 256))
filas += [('densa 8192→128', 'k_matmul', '(8, 4)', '16×16 = 256', 'baldosas 16×16 en memoria compartida', 8 * 4 * 256),
          ('softmax + pérdida', 'k_softmax_entropia', '(1,)', '64', '1 imagen por hilo', 64),
          ('Adam (conv3_w)', 'k_adam', f'({-(-73728 // 256)},)', '256', '1 parámetro por hilo', 73728)]
display(pd.DataFrame(filas, columns=['Operación', 'Kernel', 'Grid (bloques)', 'Hilos por bloque',
                                     'Trabajo por hilo', 'Hilos lanzados'])
        .style.format({'Hilos lanzados': '{:,}'}))
````

**Decisiones de diseño de los kernels (detalle en los docstrings de `src/kernels_cuda.py`):**

1. **Convolución como producto de matrices implícito (*implicit GEMM*).** La convolución se reescribe como `salida(P×C_out) = A(P×9·C_in) · W(9·C_in×C_out)`, con P = imágenes × alto × ancho, donde cada fila de A son los 3×3×C_in vecinos de un píxel. A nunca se guarda en memoria (ocuparía 9 veces la entrada): cada bloque calcula su parte al cargarla en memoria compartida. Los pesos en formato HWIO de Keras ya tienen exactamente la forma de W.
2. **Memoria compartida.** Cada bloque carga baldosas de A y de W en memoria compartida (`cuda.shared.array`) y cada valor se reutiliza 16–64 veces antes de volver a la memoria global.
3. **Register tiling.** En `k_conv_gemm_rt` cada hilo acumula 4×4 salidas en registros: por cada valor leído de memoria compartida hace 4 multiplicaciones-suma en lugar de 1.
4. **Split-K con atomics** para el gradiente de los pesos: la reducción recorre hasta 262,144 píxeles y la salida es pequeña (27×32 en la primera capa), así que la suma se reparte en trozos (`blockIdx.z`) que se combinan con `cuda.atomic.add`.
5. **El gradiente de la entrada es otra convolución** con los pesos volteados 180° y transpuestos (`k_voltear_pesos`), así que reutiliza el mismo kernel.
6. **Sin atomics en el pooling:** las ventanas 2×2 no se solapan, así que la retropropagación es un *gather* (un hilo por píxel de entrada) y se fusiona con la ReLU.
7. **Solo float32:** la MX450 calcula float64 a 1/32 de la velocidad de float32. Todas las constantes son `float32` para que Numba no promueva a float64.
8. **Kernels cortos:** en Windows (modo WDDM) un kernel de más de ~2 s reinicia el controlador (TDR); por eso se trabaja lote por lote.
9. **Doble buffer y copias asíncronas:** la CPU arma el lote i+1 en memoria *pinned* mientras la GPU calcula el lote i.

### 5.3 Kernels propios en CUDA Python (Numba)

**Correctitud.** Se compara una pasada completa (adelante y atrás, sin dropout) contra una referencia en PyTorch con los mismos pesos. Deben coincidir los logits, la pérdida y los 10 gradientes.

````python
import torch.nn.functional as F
from numba import cuda

N = 64
rng = np.random.default_rng(0)
Xr, yr = X_train[:N], y_train[:N].astype(np.int32)
m = kc.CNNBaseNumba(lote=N, seed=42, tasa_dropout=0.0)
m.cargar_lote(Xr, yr); m.forward(entrenando=True); m.backward(); cuda.synchronize()

P = {k: torch.tensor(v, requires_grad=True) for k, v in kc.pesos_iniciales(42).items()}
x = torch.tensor(Xr).permute(0, 3, 1, 2)
for i in (1, 2, 3):
    x = F.max_pool2d(F.relu(F.conv2d(x, P[f'conv{i}_w'].permute(3, 2, 0, 1), P[f'conv{i}_b'], padding=1)), 2)
logits = F.relu(x.permute(0, 2, 3, 1).reshape(N, -1) @ P['dense1_w'] + P['dense1_b']) @ P['dense2_w'] + P['dense2_b']
perdida = F.cross_entropy(logits, torch.tensor(yr, dtype=torch.long)); perdida.backward()

err = lambda a, b: float(np.abs(a - b).max() / (np.abs(b).max() + 1e-12))
filas = [('logits', err(m.logits.copy_to_host(), logits.detach().numpy())),
         ('pérdida', abs(m.perdidas.copy_to_host().mean() - perdida.item()))]
filas += [(f'∂L/∂{k}', err(m.grads[k].copy_to_host(), P[k].grad.numpy())) for k in P]
display(pd.DataFrame(filas, columns=['Tensor', 'Error relativo máximo vs PyTorch']).style.format({'Error relativo máximo vs PyTorch': '{:.1e}'}))
````

**Optimización paso a paso.** Se miden las tres versiones de los kernels de convolución sobre el mismo lote (20 pasos completos de entrenamiento cada una). Esta celda corre justo después del entrenamiento completo, con la GPU caliente (~74 °C) y su reloj limitado a ~450 MHz: los tiempos absolutos son los de una GPU frenada, pero la mejora entre versiones es la que importa.

````python
def ms_por_paso(implementacion, repeticiones=20):
    modelo = kc.CNNBaseNumba(lote=64, seed=42, implementacion=implementacion)
    modelo.cargar_lote(X_train[:64], y_train[:64].astype(np.int32))
    modelo.paso_entrenamiento(); cuda.synchronize()          # compilación JIT
    t0 = time.perf_counter()
    for _ in range(repeticiones):
        modelo.paso_entrenamiento()
    cuda.synchronize()
    return (time.perf_counter() - t0) / repeticiones * 1000

versiones = {'v1 directa (un hilo por salida)': 'directa',
             'v2 implicit GEMM, baldosas 16×16': 'gemm',
             'v3 implicit GEMM + register tiling 4×4': 'gemm_rt'}
with MonitorRecursos('resultados/monitor/notebook_numba_versiones.csv', 'numba_versiones'):
    tiempos_v = {k: ms_por_paso(v) for k, v in versiones.items()}
tabla_v = pd.DataFrame({'ms por paso (lote 64)': tiempos_v})
tabla_v['s por época (254 pasos)'] = tabla_v['ms por paso (lote 64)'] * 254 / 1000
tabla_v['Mejora sobre v1'] = tabla_v['ms por paso (lote 64)'].iloc[0] / tabla_v['ms por paso (lote 64)']
tabla_v.to_csv('resultados/notebook_numba_versiones.csv')
display(tabla_v.style.format(precision=1).format({'Mejora sobre v1': '{:.1f}×'}))
````

Las corridas cortas (CNN Base en PyTorch, ~25 s cada una) se midieron casi a reloj completo; las largas (CNN Profunda, aumento de datos y los kernels de Numba, varios minutos) a un tercio del reloj. Por eso el speedup de la CNN Base en PyTorch es optimista para un entrenamiento largo, y el de Numba es pesimista frente a su velocidad a reloj completo (sección 5.3).

**El CPU también se frena.** En la CNN Base (corridas de ~100 s con pausas de 60 s), Keras tardó ~35 s por época, como en el Lab 3 completo (~30 s). En la CNN Profunda (12 minutos seguidos por corrida), la primera época empezó a ~600 ms por paso, como en el Lab 3. A los ~2 minutos subió a ~955 ms por paso y ahí se quedó: ~240 s por época contra ~147 s del Lab 3 completo. El CPU siguió ocupado al 97 %, así que corría a menor frecuencia. No hubo escaneos del antivirus ni cambios del modo de energía, y la RAM estuvo en 13.4 GB. La causa es el control térmico de Intel (DPTF), que no se activó en el Lab 3 completo, corrido en otra sesión. Por eso el speedup de los modelos profundos (~7×) es optimista. Contra el tiempo por época del Lab 3 completo, la CNN Profunda se acelera ~4.3× y la CNN Base ~3.4×.

### 6.2 Speedup por fase y de extremo a extremo

````python
par8 = res_prep[(res_prep.variante == 'paralela') & (res_prep.procesos == 8)].iloc[0]
t_prep_seq, t_prep_par = sec.media_s, par8.media_s
t_train_cpu = completo['Minutos CPU (Lab 3, esta laptop)'].sum() * 60
t_train_gpu = completo['Minutos GPU'].sum() * 60
inf = res_train[res_train.framework.isin(['keras', 'torch'])].pivot_table(
    index='modelo', columns=['framework', 'dispositivo'], values='inferencia_s')
t_inf_cpu, t_inf_gpu = inf[('keras', 'cpu')].sum(), inf[('torch', 'cuda')].sum()

fases = pd.DataFrame({
    'Original (s)': [t_prep_seq, t_train_cpu, t_inf_cpu],
    'Paralelo (s)': [t_prep_par, t_train_gpu, t_inf_gpu],
    'Hardware paralelo': ['CPU, 8 procesos', 'GPU MX450 (PyTorch + cuDNN)', 'GPU MX450'],
}, index=['1. Preparación de 23,200 imágenes', '2. Entrenamiento de los 4 modelos (EarlyStopping)',
          '3. Inferencia sobre prueba (4 modelos)'])
fases.loc['Total del pipeline'] = [fases['Original (s)'].sum(), fases['Paralelo (s)'].sum(), '']
fases['Speedup'] = fases['Original (s)'] / fases['Paralelo (s)']
fases['% del tiempo original'] = fases['Original (s)'] / fases.loc['Total del pipeline', 'Original (s)']
fases.to_csv('resultados/speedup_por_fase.csv')
display(fases.style.format({'Original (s)': '{:,.1f}', 'Paralelo (s)': '{:,.1f}', 'Speedup': '{:.2f}×',
                            '% del tiempo original': '{:.1%}'}))
````

**Nota sobre la fila de entrenamiento:** con EarlyStopping, CPU y GPU pueden detenerse en épocas distintas (el orden de las operaciones en punto flotante cambia y con él la trayectoria del entrenamiento). Por eso el speedup **por época** de la tabla 6.1 es la medida limpia del hardware. El total de esta tabla es la medida práctica: cuánto esperó el equipo.

**Ley de Amdahl.** Si una fracción *f* del tiempo se acelera *s* veces, el speedup total es 1 / ((1 − f) + f / s). El entrenamiento es casi todo el tiempo original, así que el speedup total queda muy cerca del speedup del entrenamiento: paralelizar la preparación casi no mueve el total en este pipeline, aunque individualmente sea la fase con mejor escalamiento.

````python
f_train = fases.loc['2. Entrenamiento de los 4 modelos (EarlyStopping)', '% del tiempo original']
s_train = fases.loc['2. Entrenamiento de los 4 modelos (EarlyStopping)', 'Speedup']
s_prep = fases.loc['1. Preparación de 23,200 imágenes', 'Speedup']
f_prep = fases.loc['1. Preparación de 23,200 imágenes', '% del tiempo original']
print(f"Solo GPU (preparación secuencial): {1 / ((1 - f_train) + f_train / s_train):.2f}×")
print(f"Solo CPU multiproceso (entrenamiento en CPU): {1 / ((1 - f_prep) + f_prep / s_prep):.3f}×")
print(f"Ambas: {fases.loc['Total del pipeline', 'Speedup']:.2f}×")
print(f"Límite si el entrenamiento fuera instantáneo: {1 / (1 - f_train):.0f}×")
````

### 6.4 Discusión técnica

*(El texto completo, con las cifras finales, está en `informe/Informe.pdf`; aquí va el resumen.)*

**¿Por qué procesar las imágenes con los núcleos del CPU y delegar solo la matemática de la red a la GPU?**
Porque cada fase tiene la forma de trabajo de un procesador distinto. Decodificar un JPEG es lógica secuencial con muchas ramas (decodificación Huffman, acceso a disco, formatos y metadatos), y la hace Pillow, una biblioteca de CPU. Una GPU ejecuta grupos de 32 hilos en el mismo paso (warps) y pierde rendimiento cuando cada hilo toma una rama distinta. Además, los archivos están en el disco del CPU: subir JPEG comprimidos para decodificarlos en la GPU agregaría transferencias y no hay un decodificador de PIL para GPU. Las convoluciones, en cambio, son millones de multiplicaciones-suma idénticas e independientes, exactamente el patrón SIMT de la GPU. Separando así, cada lado hace lo que hace mejor y por el bus PCIe solo viajan tensores listos (3 MB por lote), no archivos.

**¿Cómo cambió CUDA el tiempo de procesamiento?**
Con cuDNN, la GPU redujo el tiempo por época de todos los modelos (tabla 6.1). El efecto es mayor cuanto más cómputo hay por imagen: la CNN Profunda (155 M de multiplicaciones-suma por imagen) gana más que la CNN Base (42 M), incluso con la referencia conservadora del Lab 3 completo (~4.3× contra ~3.4×), porque con poco trabajo por lote el costo fijo (lanzar kernels, copiar el lote, validar) pesa más. Las transferencias CPU→GPU resultaron una fracción pequeña del tiempo de cada época (tabla 6.3), gracias a la memoria *pinned* y a lotes de solo 3 MB. Los kernels propios en Numba quedan detrás de cuDNN: con la GPU fría la v3 tardó ~68 ms por lote (~17 s por época), menos que Keras en CPU; pero con la GPU caliente, a ~400–450 MHz, el paso sube a ~175 ms (tabla de la sección 5.3) y la época de un entrenamiento real a ~53 s, más que Keras con oneDNN y AVX-512 en los 8 hilos (tabla 6.1), aunque 2.4 veces más rápido que PyTorch en CPU. La diferencia con cuDNN es el trabajo de optimización que NVIDIA invierte en cada arquitectura (selección de algoritmo, Winograd, kernels ajustados a mano).

**¿Qué factores justifican o limitan CUDA?**
*Lo justifican* el volumen de operaciones aritméticas independientes, el tamaño de lote suficiente para llenar los multiprocesadores, y que los datos se queden en la GPU entre operaciones. *Lo limitan*, en este caso: (1) una GPU de entrada (MX450: 896 núcleos, 2 GB) frente a una CPU fuerte con AVX-512; (2) el **límite de potencia de la laptop**: tras ~30 s de carga, la GPU baja su reloj de ~1,400 MHz a ~400–500 MHz (`nvidia-smi` reporta `SW Power Cap`), un tercio de su capacidad; (3) memoria de 2 GB, que obliga a transferir por lotes; (4) la Ley de Amdahl: lo que no corre en la GPU acota el speedup total; (5) que TensorFlow no tenga soporte de GPU en Windows nativo, lo que obligó a cambiar a PyTorch o usar WSL2; (6) la compatibilidad de versiones: el NVVM de Numba tuvo que fijarse a la misma versión 12.8 del nvJitLink que trae PyTorch.

**Bibliotecas que permitieron usar la GPU:** PyTorch (`torch.cuda`, tensores en GPU, `pin_memory`, `torch.cuda.Event`, `torch.cuda.synchronize`), cuDNN (convoluciones, BatchNorm y pooling, con `cudnn.benchmark`), cuBLAS (capas densas, a través de PyTorch) y Numba CUDA (`numba.cuda.jit`, `cuda.shared.array`, `cuda.local.array`, `cuda.atomic.add`, `cuda.syncthreads`, `cuda.event`, `cuda.pinned_array`, `numba.cuda.random`). En CPU: `concurrent.futures.ProcessPoolExecutor` y `multiprocessing.shared_memory`. Monitoreo: `psutil` y `nvidia-smi`.
